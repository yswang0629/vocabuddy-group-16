```python
import random

vocabulary = [
    {
        "english": "algorithm",
        "part_of_speech": "n.",
        "chinese": "演算法"
    },
    {
        "english": "recursion",
        "part_of_speech": "n.",
        "chinese": "遞迴"
    },
    {
        "english": "efficient",
        "part_of_speech": "adj.",
        "chinese": "有效率的"
    },
]

def show_random_vocabulary():
    word = random.choice(vocabulary)
    print(f"{word['english']} ({word['part_of_speech']}) {word['chinese']}")

show_random_vocabulary()
```
