```python
import random
import unicodedata

vocabulary = [
    {"english": "algorithm", "part_of_speech": "n.", "chinese": "演算法"},
    {"english": "recursion", "part_of_speech": "n.", "chinese": "遞迴"},
    {"english": "efficient", "part_of_speech": "adj.", "chinese": "有效率的"},
    {"english": "collaborate", "part_of_speech": "v.", "chinese": "合作"},
    {"english": "curious", "part_of_speech": "adj.", "chinese": "好奇的"},
    {"english": "challenge", "part_of_speech": "n.", "chinese": "挑戰"},
    {"english": "sustainable", "part_of_speech": "adj.", "chinese": "永續的"},
    {"english": "recycle", "part_of_speech": "v.", "chinese": "回收"},
    {"english": "exercise", "part_of_speech": "n.", "chinese": "運動"},
    {"english": "confident", "part_of_speech": "adj.", "chinese": "有自信的"},
    {"english": "analyze", "part_of_speech": "v.", "chinese": "分析"},
    {"english": "creative", "part_of_speech": "adj.", "chinese": "有創意的"},
    {"english": "decision", "part_of_speech": "n.", "chinese": "決定"},
    {"english": "communicate", "part_of_speech": "v.", "chinese": "溝通"},
    {"english": "healthy", "part_of_speech": "adj.", "chinese": "健康的"},
    {"english": "technology", "part_of_speech": "n.", "chinese": "科技"},
    {"english": "protect", "part_of_speech": "v.", "chinese": "保護"},
    {"english": "responsible", "part_of_speech": "adj.", "chinese": "負責任的"},
    {"english": "opportunity", "part_of_speech": "n.", "chinese": "機會"},
    {"english": "improve", "part_of_speech": "v.", "chinese": "改善"},
    {"english": "anachronistic", "part_of_speech": "adj.", "chinese": "過時的"},
    {"english": "substantiate", "part_of_speech": "v", "chinese": "證實"},
    {"english": "conundrum", "part_of_speech": "n.", "chinese": "難題"},
    {"english": "capricious", "part_of_speech": "adj.", "chinese": "反覆無常的"},
    {"english": "establish", "part_of_speech": "v.", "chinese": "建立"},
    {"english": "phenomenon", "part_of_speech": "n.", "chinese": "現象"},
    {"english": "contemporary", "part_of_speech": "adj.", "chinese": "當代的"},
    {"english": "demonstrate", "part_of_speech": "v.", "chinese": "證明"},
    {"english": "perspective", "part_of_speech": "n.", "chinese": "看法"},
    {"english": "deleterious", "part_of_speech": "adj.", "chinese": "有害的"}
]

correct_count = 0
wrong_count = 0
answer_records = []


def display_width(text):
    return sum(
        2 if unicodedata.east_asian_width(char) in "WFA" else 1
        for char in text
    )


def pad_display(text, width):
    return text + " " * max(0, width - display_width(text))


def show_total():
    print("═" * 40)
    print("       VocaBuddy 單字測驗")
    print("═" * 40)
    print(f"  目前共有 {len(vocabulary)} 個單字。")


def vocabulary_quiz(correct_word, question_number):
    global correct_count, wrong_count

    # 四個選項的詞性都與題目相同
    same_part_words = [
        word for word in vocabulary
        if word["part_of_speech"] == correct_word["part_of_speech"]
        and word != correct_word
    ]

    wrong_words = random.sample(same_part_words, 3)
    choices = wrong_words + [correct_word]
    random.shuffle(choices)

    labels = ["A", "B", "C", "D"]
    correct_label = labels[choices.index(correct_word)]

    print("\n" + "═" * 25)
    print(f"  第 {question_number} 題／共 5 題")
    print("═" * 25)
    print(f"  英文：{correct_word['english']}")
    print("─" * 25)
    print("  選項    中文解釋         詞性")
    print("─" * 25)

    for label, word in zip(labels, choices):
        chinese = pad_display(word["chinese"], 16)
        print(f"  {label}.      {chinese} {word['part_of_speech']}")

    print("─" * 25)
    answer = input("  請輸入答案（A、B、C 或 D）：").strip().upper()

    if answer == correct_label:
        result = "答對 ✓"
        correct_count += 1
        print(f"\n  ✓ 答對！{correct_word['english']} = {correct_word['chinese']}")
    else:
        result = "答錯 ✗"
        wrong_count += 1
        print(f"\n  ✗ 答錯。正確答案是 {correct_label}. {correct_word['chinese']}")

    answer_records.append({
        "word": correct_word["english"],
        "your_answer": answer,
        "correct_answer": correct_label,
        "result": result
    })


def show_records():
    print("\n" + "═" * 58)
    print("                     測驗答題紀錄")
    print("═" * 58)
    print("  題號  英文單字              你的答案  正解  結果")
    print("─" * 58)

    for number, record in enumerate(answer_records, start=1):
        print(
            f"  {number:<4}  "
            f"{record['word']:<20}  "
            f"{record['your_answer']:<8}  "
            f"{record['correct_answer']:<4}  "
            f"{record['result']}"
        )

    total = correct_count + wrong_count

    print("═" * 28)
    print(f"  總題數：{total} 題｜答對：{correct_count} 題｜答錯：{wrong_count} 題")
    print(f"  正確率：{correct_count / total * 100:.0f}%")
    print("═" * 28)


show_total()

quiz_words = random.sample(vocabulary, 5)

for question_number, word in enumerate(quiz_words, start=1):
    vocabulary_quiz(word, question_number)

show_records()
```
